```python
ZWNJ = "\u200c"

string = 'fsddavdsvsdvsdvsvsv'
async def NormalizeData(s) :
    if not string:
        return ""
    s = s.replace("\u064a", "ی").replace("\u0649", "ی").replace("\u06cc", "ی")
    s = s.replace("\u0643", "ک").replace("\u06a9", "ک") 
    s = s.replace(ZWNJ, " ")
    s = re.sub(r"[\u064B-\u0652]", "", s)
    s = s.strip()
    s = re.sub(r"\s+", " ", s)
    return s
```
